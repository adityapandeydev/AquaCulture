# DO experiment setup

Inspect the approved data and record a development run through the shared Python modules. This notebook fits no model and does not expose final-test or calibration batches. Use the model-environment kernel described in `docs/models/development.md`.

```python
from pathlib import Path
import sys

here = Path.cwd().resolve()
repository = next(path for path in [here, *here.parents] if (path / 'docs/data/prepared-release.json').is_file())
if str(repository) not in sys.path:
    sys.path.insert(0, str(repository))

from ml.training.experiment import inspect_experiment
```

```python
from datetime import datetime, timezone

run_id = datetime.now(timezone.utc).strftime('foundation-notebook-%Y%m%dT%H%M%S%fZ')
report = inspect_experiment(
    repository / 'ml/training/configs/do_pilot.json',
    repository / 'ml/runs' / run_id,
    repository=repository,
)
report['development_partitions']
```

After installing the CUDA-enabled model environment, run `python -m ml.training.experiment doctor --device cuda` in its terminal.
